ئەڤە ب تەمامی بلاشە و تۆمارکرن ناڤێت. دەنگ دەنگێ Meta MMS (Kurmancî) ـە.

⚠️ رستان ب **پیتێن لاتینی** بنڤیسە: `Silav, ez Zîrek im.`

```
#@title 1. ئامادەکرن (١–٢ خولەک)
!pip -q install transformers
import torch
from transformers import VitsModel, AutoTokenizer
from IPython.display import Audio, display

NAME = "facebook/mms-tts-kmr-script_latin"
model = VitsModel.from_pretrained(NAME)
tokenizer = AutoTokenizer.from_pretrained(NAME)

def say(text):
    inputs = tokenizer(text, return_tensors="pt")
    with torch.no_grad():
        wav = model(**inputs).waveform[0].numpy()
    display(Audio(wav, rate=model.config.sampling_rate, autoplay=True))

print("Amade ye ✅")
```

```
#@title 2. رستا خۆ بنڤیسە و ▶️ بکە
rist = "Silav, ez Zîrek im. Ez robotek im û bi kurdiya Badînî diaxivim." #@param {type:"string"}
lez = 1.0 #@param {type:"slider", min:0.6, max:1.4, step:0.1}
model.speaking_rate = lez
say(rist)
```

```
#@title 3. چەند رستێن نموونە
for s in [
    "Sibeha te bi xêr, tu çawa yî?",
    "Îro hewa sayî ye û roj gelek germ e.",
    "Duhok bajarekê xoşik e di nav çiyayan da.",
    "Spas bo alîkariya te, şev baş.",
]:
    print(s)
    say(s)
```
